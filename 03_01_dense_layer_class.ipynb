```python
import numpy as np
from nnfs.datasets import spiral_data
```

```python
class Layer_Dense:
    # n_inputs: number of input features or number of neurons in the preceding layer which are feeding into this layer
    # n_neurons: number of neurons in the current layer
    def __init__(self, n_inputs, n_neurons):
        self.weights = np.random.randn(n_inputs, n_neurons)
        self.biases = np.zeros((1, n_neurons))

    def forward_pass(self, inputs):
        self.output = np.dot(inputs, self.weights) + self.biases
```

```python
# create dataset
X, y = spiral_data(samples=100, classes=3)

# create Dense Layer with 2 input features and 3 output values
dense1 = Layer_Dense(2, 3)

# perform a forward pass of our training data through this layer
dense1.forward_pass(X)

# Print the output for first few samples:
print(dense1.output[:5])
```
