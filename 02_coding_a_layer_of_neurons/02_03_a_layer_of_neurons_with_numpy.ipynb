```python
import numpy as np
```

```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights = [
    [0.2, 0.8, -0.5, 1.0],
    [0.5, -0.91, 0.26, -0.5],
    [-0.26, -0.27, 0.17, 0.87]
]
bias = [2.0, 3.0, 0.5]
```

```python
layer_outputs = []
for neuron_weights, neuron_bias in zip(weights, bias):
    layer_outputs.append(np.dot(inputs, neuron_weights) + neuron_bias)

print(np.round(layer_outputs, 3))
```
