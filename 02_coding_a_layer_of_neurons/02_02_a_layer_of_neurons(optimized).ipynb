```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights = [
    [0.2, 0.8, -0.5, 1.0],
    [0.5, -0.91, 0.26, -0.5],
    [-0.26, -0.27, 0.17, 0.87]
]
bias = [2.0, 3.0, 0.5]
```

```python
print(list(zip(weights, bias)))
```

```python
layer_outputs = []
for neuron_weights, neuron_bias in zip(weights, bias):
    layer_outputs.append(sum([x * y for x, y in zip(inputs, neuron_weights)]) + neuron_bias)
print(layer_outputs)
```
