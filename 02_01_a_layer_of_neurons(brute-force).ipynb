```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights_of_neuron1 = [0.2, 0.8, -0.5, 1.0]
weights_of_neuron2 = [0.5, -0.91, 0.26, -0.5]
weights_of_neuron3 = [-0.26, -0.27, 0.17, 0.87]
bias_of_neuron1 = 2.0
bias_of_neuron2 = 3.0
bias_of_neuron3 = 0.5
```

```python
output = [
    # neuron 1
    inputs[0] * weights_of_neuron1[0] + 
    inputs[1] * weights_of_neuron1[1] + 
    inputs[2] * weights_of_neuron1[2] + 
    inputs[3] * weights_of_neuron1[3] + bias_of_neuron1,
    # neuron 2
    inputs[0] * weights_of_neuron2[0] + 
    inputs[1] * weights_of_neuron2[1] + 
    inputs[2] * weights_of_neuron2[2] + 
    inputs[3] * weights_of_neuron2[3] + bias_of_neuron2,
    # neuron 3
    inputs[0] * weights_of_neuron3[0] + 
    inputs[1] * weights_of_neuron3[1] + 
    inputs[2] * weights_of_neuron3[2] + 
    inputs[3] * weights_of_neuron3[3] + bias_of_neuron3
]
```

```python
print(output)
```
