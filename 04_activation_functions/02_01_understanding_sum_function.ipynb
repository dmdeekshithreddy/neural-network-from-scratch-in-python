## Understanding the sum function

````python
import numpy as np
````

````python
layer_outputs = np.array([[4.8, 1.21, 2.385],
[8.9, -1.81, 0.2],
[1.41, 1.051, 0.026]])

# sum of all the elements
print(np.sum(layer_outputs))

# sum the elements row-wise ; here row-wise meaning moving along the rows; take 1 element from each row
print(np.sum(layer_outputs, axis=0)) # 4.8 + 8.9 + 1.41 = 15.11

# sum the elements column-wise; take 1 element from each column
print(np.sum(layer_outputs, axis=1)) # 4.8 + 1.21 + 2.385 = 8.395
````

### Notes on `axis`:
- `axis=0` means "collapse the rows" (sum down each column)
- `axis=1` means "collapse the columns" (sum across each row)

You're right to be confused: "row-wise" is ambiguous. Use this rule instead: **`axis` names the dimension that disappears.**

#### Part 1: `axis=0` gets rid of the rows

```
            col 0    col 1    col 2
row 0  [  4.8,    1.21,   2.385 ]
row 1  [  8.9,   -1.81,   0.2   ]
row 2  [  1.41,   1.051,  0.026 ]
            ↓        ↓        ↓
axis=0  [ 15.11,   0.451,  2.611 ]   ← one number per column
```

NumPy moves down through the rows and adds them, so the three rows collapse into one. What's left is one total per column.

Your comment `4.8 + 8.9 + 1.41 = 15.11` is correct. That is the total of **column 0**. A clearer comment:

```python
print(np.sum(layer_outputs, axis=0))   # sum down each column → [15.11, 0.451, 2.611]
```

#### Part 2: `axis=1` gets rid of the columns

```
row 0  [ 4.8,  1.21,  2.385 ]  →  8.395
row 1  [ 8.9, -1.81,  0.2   ]  →  ?
row 2  [ 1.41, 1.051, 0.026 ]  →  2.487
```

NumPy moves across each row and adds the values, so the columns collapse. What's left is one total per row.

#### Part 3: The same thing in SQL

| NumPy | SQL equivalent | Result |
|---|---|---|
| `axis=0` | `SELECT SUM(c0), SUM(c1), SUM(c2) FROM t` | one value per column |
| `axis=1` | `SELECT c0 + c1 + c2 FROM t` | one value per row |
| no axis | sum of everything | one number: 18.172 |

#### Why the word "row-wise" confuses people

People use it in two opposite ways:

- "Moving along the rows" means going down, which is `axis=0`.
- "For each row" means going across, which is `axis=1`.

So skip the word. Ask yourself which dimension should disappear.

In neural network code you'll mostly use `axis=1`. Each row is one sample, and you want one result per sample. Softmax, for example, adds up each row's values.

### Notes on keepdims

````python
print(np.sum(layer_outputs, axis=1))
print(np.sum(layer_outputs, axis=1).shape)

print(np.sum(layer_outputs, axis=1, keepdims=True))
print(np.sum(layer_outputs, axis=1, keepdims=True).shape)
````
